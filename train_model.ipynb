# Carbon Emission Model Training

This notebook trains the model used by the Carbon Emission Estimator app. The workflow loads the supply-chain dataset, cleans cost values, adds useful sustainability features, compares regression models, and saves the best pipeline.

```python
import re
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.compose import ColumnTransformer, TransformedTargetRegressor
from sklearn.dummy import DummyRegressor
from sklearn.ensemble import ExtraTreesRegressor, GradientBoostingRegressor, HistGradientBoostingRegressor, RandomForestRegressor
from sklearn.impute import SimpleImputer
from sklearn.linear_model import ElasticNet, LinearRegression, RidgeCV
from sklearn.metrics import mean_absolute_error, r2_score, root_mean_squared_error
from sklearn.model_selection import KFold, cross_validate, train_test_split
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler

SEED = 42
DATA_PATH = "SCM_Dataset_Updated_with_Green_Logistics.xlsx"
MODEL_PATH = "carbon_emissions_model.joblib"
TARGET = "Carbon Emissions (kg CO2e)"
ROW_IDENTIFIERS = ["Company Name"]

sns.set_theme(style="whitegrid")
```

## Load Dataset

The dataset contains supply-chain, operation, sustainability, technology, and risk details. The target column is total carbon emissions in kg CO2e.

```python
df = pd.read_excel(DATA_PATH)
df.head()
```

```python
df.info()
df.describe(include="all").T
```

```python
def parse_compact_money(raw_value):
    """Convert compact money labels such as '$700B' or '$12.5M' to numbers."""
    if pd.isna(raw_value):
        return np.nan

    cleaned = str(raw_value).strip().replace(",", "").replace("$", "")
    match = re.fullmatch(r"([-+]?\d*\.?\d+)\s*([KMBT]?)", cleaned, flags=re.IGNORECASE)
    if match is None:
        return pd.to_numeric(cleaned, errors="coerce")

    amount = float(match.group(1))
    scale = {"": 1, "K": 1e3, "M": 1e6, "B": 1e9, "T": 1e12}[match.group(2).upper()]
    return amount * scale


def enrich_supply_chain_frame(frame):
    """Add simple sustainability and efficiency features."""
    enriched = frame.copy()
    supplier_count = enriched["Supplier Count"].replace(0, np.nan)
    energy = enriched["Energy Consumption (MWh)"]
    renewable_ratio = enriched["Use of Renewable Energy (%)"] / 100
    recycling_ratio = enriched["Recycling Rate (%)"] / 100
    packaging_ratio = enriched["Green Packaging Usage (%)"] / 100

    enriched["Energy per Supplier"] = energy / supplier_count
    enriched["Implementation Cost per Supplier"] = enriched["Total Implementation Cost"] / supplier_count
    enriched["Renewable Energy MWh"] = energy * renewable_ratio
    enriched["Non Renewable Energy MWh"] = energy * (1 - renewable_ratio)
    enriched["Recycling Packaging Score"] = recycling_ratio * enriched["Green Packaging Usage (%)"]
    enriched["Efficiency Risk Balance"] = enriched["Operational Efficiency Score"] - enriched["Supply Chain Risk (%)"]
    enriched["Log Energy Consumption"] = np.log1p(energy)
    enriched["Circularity Index"] = (recycling_ratio + packaging_ratio + renewable_ratio) / 3
    enriched["Supplier Energy Load"] = np.log1p(enriched["Energy per Supplier"])
    enriched["Fulfillment Efficiency Gap"] = (100 - enriched["Order Fulfillment Rate (%)"]) + (100 - enriched["Operational Efficiency Score"])
    return enriched


data = df.copy()
data["Cost of Goods Sold (COGS)"] = data["Cost of Goods Sold (COGS)"].apply(parse_compact_money)
data = data.dropna(subset=[TARGET]).drop_duplicates()
data = enrich_supply_chain_frame(data)

feature_drop_cols = [TARGET, *ROW_IDENTIFIERS]
X = data.drop(columns=feature_drop_cols)
y = data[TARGET].astype(float)

engineered_features = [
    "Energy per Supplier",
    "Implementation Cost per Supplier",
    "Renewable Energy MWh",
    "Non Renewable Energy MWh",
    "Recycling Packaging Score",
    "Efficiency Risk Balance",
    "Log Energy Consumption",
    "Circularity Index",
    "Supplier Energy Load",
    "Fulfillment Efficiency Gap",
]

print(f"Rows used for modeling: {len(data):,}")
print(f"Feature columns: {X.shape[1]}")
print(f"Target range: {y.min():,.0f} to {y.max():,.0f} kg CO2e")
print("Added features:", engineered_features)
```

## Split and Preprocessing Design

The split happens before fitting imputers, scalers, or encoders. That keeps the holdout set closer to how the app will behave with new scenarios.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    random_state=SEED,
)

numeric_features = X_train.select_dtypes(include=["number", "bool"]).columns.tolist()
categorical_features = X_train.select_dtypes(exclude=["number", "bool"]).columns.tolist()

numeric_pipeline = Pipeline(
    steps=[
        ("fill_missing", SimpleImputer(strategy="median")),
        ("scale", StandardScaler()),
    ]
)

categorical_pipeline = Pipeline(
    steps=[
        ("fill_missing", SimpleImputer(strategy="most_frequent")),
        ("encode", OneHotEncoder(handle_unknown="ignore", sparse_output=False)),
    ]
)

preprocessor = ColumnTransformer(
    transformers=[
        ("numeric", numeric_pipeline, numeric_features),
        ("categorical", categorical_pipeline, categorical_features),
    ]
)

print("Numeric features:", numeric_features)
print("Categorical features:", categorical_features)
```

## Compare Models

Several regression models are compared. The saved model includes preprocessing, so the app can make predictions directly.

```python
def model_pipeline(regressor, log_target=False):
    base_pipeline = Pipeline(
        steps=[
            ("preprocess", preprocessor),
            ("regressor", regressor),
        ]
    )

    if not log_target:
        return base_pipeline

    return TransformedTargetRegressor(
        regressor=base_pipeline,
        func=np.log1p,
        inverse_func=np.expm1,
    )


def model_zoo():
    return {
        "Baseline mean": DummyRegressor(strategy="mean"),
        "Linear regression": model_pipeline(LinearRegression()),
        "Ridge regression": model_pipeline(RidgeCV(alphas=np.logspace(-8, 4, 40))),
        "ElasticNet": model_pipeline(ElasticNet(alpha=0.001, l1_ratio=0.05, max_iter=20_000, random_state=SEED)),
        "Extra trees": model_pipeline(
            ExtraTreesRegressor(
                n_estimators=800,
                min_samples_leaf=1,
                max_features=1.0,
                random_state=SEED,
                n_jobs=-1,
            )
        ),
        "Random forest": model_pipeline(
            RandomForestRegressor(
                n_estimators=800,
                min_samples_leaf=1,
                max_features=1.0,
                random_state=SEED,
                n_jobs=-1,
            )
        ),
        "Gradient boosting": model_pipeline(
            GradientBoostingRegressor(
                n_estimators=600,
                learning_rate=0.03,
                max_depth=3,
                subsample=0.9,
                random_state=SEED,
            )
        ),
        "Hist gradient boosting log target": model_pipeline(
            HistGradientBoostingRegressor(
                learning_rate=0.04,
                max_iter=700,
                l2_regularization=0.01,
                random_state=SEED,
            ),
            log_target=True,
        ),
    }


models = model_zoo()
scoring = {
    "mae": "neg_mean_absolute_error",
    "rmse": "neg_root_mean_squared_error",
    "r2": "r2",
}
cv_splits = min(5, max(2, len(X_train) // 10))
cv = KFold(n_splits=cv_splits, shuffle=True, random_state=SEED)

cv_rows = []
for name, estimator in models.items():
    scores = cross_validate(estimator, X_train, y_train, scoring=scoring, cv=cv, n_jobs=-1)
    cv_rows.append(
        {
            "model": name,
            "cv_mae": -scores["test_mae"].mean(),
            "cv_rmse": -scores["test_rmse"].mean(),
            "cv_r2": scores["test_r2"].mean(),
        }
    )

cv_results = pd.DataFrame(cv_rows).sort_values("cv_rmse")
cv_results
```

## Test Set Evaluation

Cross-validation ranks the models, then the selected model is evaluated on the test set.

```python
def evaluate_regression(y_true, y_pred, prefix="test"):
    valid = y_true != 0
    mape = np.mean(np.abs((y_true[valid] - y_pred[valid]) / y_true[valid])) * 100
    return {
        f"{prefix}_mae": mean_absolute_error(y_true, y_pred),
        f"{prefix}_rmse": root_mean_squared_error(y_true, y_pred),
        f"{prefix}_r2": r2_score(y_true, y_pred),
        f"{prefix}_mape_%": mape,
    }


test_rows = []
fitted_models = {}
for name, estimator in models.items():
    estimator.fit(X_train, y_train)
    predictions = estimator.predict(X_test)
    fitted_models[name] = estimator
    test_rows.append({"model": name, **evaluate_regression(y_test, predictions)})

test_results = pd.DataFrame(test_rows)
model_results = cv_results.merge(test_results, on="model").sort_values("cv_rmse")
model_results["cv_rank"] = model_results["cv_rmse"].rank(method="min")
model_results["test_rank"] = model_results["test_rmse"].rank(method="min")

best_model_name = model_results.iloc[0]["model"]
best_model = fitted_models[best_model_name]
test_pred = best_model.predict(X_test)
test_metrics = test_results.loc[test_results["model"] == best_model_name].iloc[0].to_dict()

model_results
```

```python
evaluation = X_test.copy()
evaluation["actual_emissions"] = y_test.values
evaluation["predicted_emissions"] = test_pred
evaluation["absolute_error"] = np.abs(evaluation["actual_emissions"] - evaluation["predicted_emissions"])
evaluation[["actual_emissions", "predicted_emissions", "absolute_error"]].sort_values("absolute_error", ascending=False).head(10)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

sns.scatterplot(x=y_test, y=test_pred, ax=axes[0])
axis_min = min(y_test.min(), test_pred.min())
axis_max = max(y_test.max(), test_pred.max())
axes[0].plot([axis_min, axis_max], [axis_min, axis_max], color="red", linestyle="--")
axes[0].set_title("Actual vs Predicted Emissions")
axes[0].set_xlabel("Actual kg CO2e")
axes[0].set_ylabel("Predicted kg CO2e")

residuals = y_test - test_pred
sns.histplot(residuals, kde=True, ax=axes[1])
axes[1].set_title("Residual Distribution")
axes[1].set_xlabel("Actual - Predicted")

plt.tight_layout()
```

## Save Model

The saved file includes the trained pipeline, feature order, feature type lists, test metrics, and model comparison results.

```python
final_model = models[best_model_name]
final_model.fit(X, y)

model_package = {
    "model": final_model,
    "target": TARGET,
    "features": X.columns.tolist(),
    "numeric_features": numeric_features,
    "categorical_features": categorical_features,
    "test_metrics": test_metrics,
    "model_results": model_results.to_dict(orient="records"),
}

joblib.dump(model_package, MODEL_PATH)
print(f"Saved {best_model_name} to {MODEL_PATH}")
```
